# ORIGIN MUSIC ENGINE — запуск ACE-Step 1.5 на бесплатном GPU Kaggle

Этот блокнот поднимает временную музыкальную студию ACE-Step 1.5 в облаке Kaggle. Он не меняет основной сайт и не требует платного API. GPU и интернет в настройках Kaggle нужно включить вручную.

## Перед запуском
1. В Kaggle откройте **Settings → Accelerator → GPU** и выберите доступный GPU.
2. В **Settings → Internet** включите интернет.
3. Запускайте ячейки по порядку. Первая установка и загрузка моделей может занять время и около 10 ГБ диска.

**Важно:** публичная ссылка Gradio временная. Не вводите в эту демо-студию пароли или личные данные. Остановите сессию Kaggle после работы, чтобы не расходовать квоту.

```python
# 1. Проверяем GPU, Python и свободное место
import sys, shutil, subprocess
print('Python:', sys.version)
print('Свободно на диске, ГБ:', round(shutil.disk_usage('/kaggle/working').free / 1024**3, 1))
try:
    print(subprocess.check_output(['nvidia-smi'], text=True))
except Exception as e:
    print('GPU не обнаружен. Проверьте Settings → Accelerator → GPU.', e)
```

```python
# 2. Загружаем официальный ACE-Step 1.5
!git clone https://github.com/ACE-Step/ACE-Step-1.5.git
%cd /kaggle/working/ACE-Step-1.5
!pip -q install uv
!uv sync
```

## Запуск музыкальной студии

Для совместимости с Tesla P100 используется PyTorch backend (`--backend pt`) и облегчённая языковая модель 0.6B. Это тестовая конфигурация; скорость и успешная загрузка зависят от фактически выданного GPU.

```python
# 3. Запускаем ACE-Step с временной ссылкой для открытия на планшете
!uv run acestep --backend pt --config_path acestep-v15-turbo --lm_model_path acestep-5Hz-lm-0.6B --offload_to_cpu --share --server-name 0.0.0.0
```

## После запуска

В выводе последней ячейки появится временная ссылка Gradio (`https://....gradio.live`). Откройте её на планшете. Не закрывайте работающую ячейку, пока пользуетесь студией.

Если Kaggle выдал T4 вместо P100, автоматическая настройка ACE-Step может позволить более мощный режим. Сначала проверим базовый запуск и качество первой композиции, а настройки будем менять после фактического теста.
